```python
#Exercise 1: Library Book Management System
class Book:
    def __init__(self, title, author, available=True):
        self.title = title
        self.author = author
        self.available = available

    def __str__(self):
        status = "Available" if self.available else "Borrowed"
        return f"{self.title} by {self.author} ({status})"


class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title, author, available=True):
        if not isinstance(title, str) or not title.strip() or title.strip().isdigit():
            print("Invalid title: must be a non-empty string.")
            return
        self.books.append(Book(title.strip(), author.strip(), available))
        print(f"Added '{title.strip()}'.")

    def borrow_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if not book.available:
                    print(f"'{book.title}' is already borrowed.")
                    return
                book.available = False
                print(f"Borrowed '{book.title}'.")
                return
        print(f"Book '{title}' not found.")

    def return_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if book.available:
                    print(f"'{book.title}' is already available.")
                    return
                book.available = True
                print(f"Returned '{book.title}'.")
                return
        print(f"Book '{title}' not found.")

    def show_books(self):
        for book in self.books:
            print("-", book)


# Sample run
lib = Library()
n = int(input("Number of books to add: "))
for _ in range(n):
    title = input("Title: ")
    author = input("Author: ")
    status = input("Available? (yes/no): ").strip().lower() == "yes"
    lib.add_book(title, author, status)

lib.show_books()
lib.borrow_book(input("Enter title to borrow: "))
lib.return_book(input("Enter title to return: "))
lib.show_books()
```

```python
#Exercise 2: Employee Payroll System
class Employee:
    def __init__(self, name, hourly_rate, hours_worked):
        self.name = name
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked

    def calculate_pay(self):
        reg_hours = min(self.hours_worked, 160)
        ot_hours = max(0, self.hours_worked - 160)
        base_salary = reg_hours * self.hourly_rate
        overtime_pay = ot_hours * self.hourly_rate * 1.5
        return base_salary, overtime_pay

    def display_summary(self):
        base, ot = self.calculate_pay()
        print(f"{self.name} | Hours: {self.hours_worked} | Base: {base:.2f} | Overtime: {ot:.2f} | Total: {base + ot:.2f}")


# Sample run
count = int(input("Number of employees: "))
while count < 1:
    count = int(input("Enter at least 1 employee: "))

employees = []
for _ in range(count):
    name = input("Name: ").strip()
    rate = float(input("Hourly wage: "))
    hours = float(input("Hours worked: "))
    if rate < 0 or hours < 0:
        print("Wage and hours cannot be negative. Skipping entry.")
        continue
    employees.append(Employee(name, rate, hours))

for emp in employees:
    emp.display_summary()
```

```python
#Exercise 3: Online Shopping Cart System
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock


class ShoppingCart:
    def __init__(self):
        self.items = {}  # {product_name: {"product": Product, "qty": int}}

    def add_item(self, product, qty):
        if qty <= 0 or qty > product.stock:
            print(f"Invalid quantity. Available stock for {product.name}: {product.stock}")
            return
        product.stock -= qty
        if product.name in self.items:
            self.items[product.name]["qty"] += qty
        else:
            self.items[product.name] = {"product": product, "qty": qty}
        print(f"Added {qty}x {product.name}.")

    def remove_item(self, name, qty):
        if name not in self.items:
            print(f"{name} is not in the cart.")
            return
        item = self.items[name]
        if qty <= 0 or qty >= item["qty"]:
            item["product"].stock += item["qty"]
            del self.items[name]
            print(f"Removed {name} from cart.")
        else:
            item["qty"] -= qty
            item["product"].stock += qty
            print(f"Removed {qty}x {name}.")

    def view_cart(self):
        total = 0
        print("\nCart:")
        for name, data in self.items.items():
            subtotal = data["product"].price * data["qty"]
            total += subtotal
            print(f"- {name} x{data['qty']}: Php.{subtotal:.2f}")
        print(f"Total: Php.{total:.2f}")


# Sample run
catalog = {
    "1": Product("Mouse", 15.0, 10),
    "2": Product("Keyboard", 45.0, 5),
    "3": Product("Headset", 30.0, 6),
}
cart = ShoppingCart()

for k, p in catalog.items():
    print(f"{k}. {p.name} (Php.{p.price:.2f}) - Stock: {p.stock}")

choice = input("Select product number to add: ")
if choice in catalog:
    cart.add_item(catalog[choice], int(input("Quantity: ")))
else:
    print("Invalid product selection.")

cart.view_cart()
```
